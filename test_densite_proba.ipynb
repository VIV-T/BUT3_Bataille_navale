```python
def est_valide(grille, row, col, taille, horizontal):
    if horizontal:
        if col + taille > len(grille[0]):
            return False
        for i in range(taille):
            if grille[row][col + i] != '-':
                if grille[row][col + i] != 'X' :
                    return False
    else:
        if row + taille > len(grille):
            return False
        for i in range(taille):
            if grille[row + i][col] != '-':
                if grille[row + i][col] != 'X' :
                    return False
    return True

def placer_navire(grille, row, col, taille, horizontal, valeur):
    if horizontal:
        for i in range(taille):
            if grille[row][col + i] != "X" :
                grille[row][col + i] = valeur
    else:
        for i in range(taille):
            if grille[row + i][col] != "X" :
                grille[row + i][col] = valeur

def retirer_navire(grille, row, col, taille, horizontal):
    if horizontal:
        for i in range(taille):
            if grille[row][col + i] != 'X' and grille[row][col + i] != '0':
                grille[row][col + i] = '-'
    else:
        for i in range(taille):
            if grille[row + i][col] != 'X' and grille[row + i][col] != '0':
                grille[row + i][col] = '-'

def configuration_valide(grille, dico_navires):
    compteur_navires = {nom[0]: 0 for nom in dico_navires}
    for row in grille:
        for cell in row:
            if cell in compteur_navires:
                compteur_navires[cell] += 1

    for nom, taille in dico_navires.items():
        if compteur_navires[nom[0]] != taille:
            return False
    return True

def generer_configurations(grille, dico_navires, dico_navires_coule=None):
    configurations = []
    if dico_navires_coule != None :
        for nom_navire in dico_navires_coule.keys() :
            dico_navires.pop(nom_navire)
    navires = list(dico_navires.items())


    def generer(index):
        if index == len(navires):
            if configuration_valide(grille, dico_navires):
                configurations.append([row[:] for row in grille])
            return

        nom, taille = navires[index]
        for row in range(len(grille)):
            for col in range(len(grille[0])):
                for horizontal in [True, False]:
                    if est_valide(grille, row, col, taille, horizontal):
                        placer_navire(grille, row, col, taille, horizontal, nom[0])
                        generer(index + 1)
                        retirer_navire(grille, row, col, taille, horizontal)

    generer(0)
    return configurations

# Exemple d'utilisation
dico_navires = {'torpilleur': 2, 'sous-marin': 3, 'cuirassé': 4}
dico_navires_coule = {'torpilleur': 2}
grille = [
    ["-", "-", "-", "-", "-"],
    ["-", "-", "-", "X", "-"],
    ["-", "0", "-", "-", "-"],
    ["-", "-", "#", "-", "-"],
    ["-", "-", "#", "-", "-"]
]

configurations = generer_configurations(grille=grille, dico_navires=dico_navires, dico_navires_coule=dico_navires_coule)
for config in configurations:
    for row in config:
        print(' '.join(row))
    print()
        
```
